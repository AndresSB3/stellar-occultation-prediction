```python
from astropy.time import Time
from astroquery.jplhorizons import Horizons
from astroquery.mpc import MPC
```

```python
# Hyperparameters
epochs = [
  {
    "start": "2020-01-01",
    "stop": "2020-02-01",
    "step": "1day",
    "number": 24
  },
  {
    "start"
  }
]
```

```python
# Para fecha individual en JPL se usa una fecha individual en formato JD, por ejemplo:
jpl_1 = Horizons(id='Ceres', epochs=Time(["2020-01-01"]).jd)

# Para lista de fechas en JPL se usa un rango de fechas en formato JD, por ejemplo:
jpl_2 = Horizons(id='Ceres', epochs=Time(["2020-01-01", "2020-01-02", "2020-01-03"]).jd)

# Para un rango de fechas en JPL se usan las normales con los campos conocidos, por ejemplo:
jpl_3 = Horizons(id='Ceres', epochs={"start": "2020-01-01", "stop": "2020-02-01", "step": "1day"})
```

```python
# Para fecha individual en MPC se asigna number como 1, por ejemplo:
mpc_1 = MPC.get_ephemeris('Ceres', start="2020-01-01", number=1, step="1day")

# Para un rango de fechas en MPC se asigna number como el número de fechas y step es relevante, por ejemplo:
mpc_2 = MPC.get_ephemeris('Ceres', start="2020-01-01", number=3, step="1day")

# **En MPC no es posible buscar una lista de fechas particulares como en JPL**
```

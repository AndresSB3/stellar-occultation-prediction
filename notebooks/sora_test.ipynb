```python
print(f'Diameter given by SORA: {body.diameter}')
print(f'With type: {type(body.diameter)}\n')

print(f'Value of diameter: {body.diameter.value}')
print(f'With type: {type(body.diameter.value)}')
```

```python
import numpy as np


def check_diameter(body):
  if body.diameter.value and not np.isnan(body.diameter.value):
    print('Diameter exists')
  else:
    print('Diameter does not exist')

check_diameter(body)
```

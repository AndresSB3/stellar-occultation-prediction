```python
import sys
from pathlib import Path

# Internal modules / packages
sys.path.append(str(Path.cwd().parent))

from occultations.core.general import run
```

```python
# Settings to test run
settings = {
  "body": {
    "id": ["Silesia"],
    "fields": {
      "object_type": "asteroid",
      "critical_list_numbered_object": None,
      "limit": 1,
      "family": None,
      "orbit_uncertainty": None,
      "state": ["is_provisional"]
    }
  },
  "database": ["JPL"],
  "epoch": {
    "start": "2019-06-27",
    "stop": "2019-06-29",
    "step": "1m",
    "number": 1000
  },
  "observer": [
    "W63"
  ],
  "limit_magnitude": 16,
  "exposition_time": 5,
  "ADS_key": None
}
```

```python
# Store predictions received by the callback
received_predictions = []

# Define callback
def test_callback(result):
  
  data = result["data"]
  table = result["table"]
  
  print("\nNew prediction received!")
  print(f"Object: {data['object']}")
  print(f"Observer: {data['observer']}")
  print(f"Prediction type: {type(data['prediction_type'])}")
  print(f"Events: {data["events"]}")
  print(f"Original table type: {type(table)}")
  
  received_predictions.append(result)
  
# Run pipeline
run(settings, prediction_callback=test_callback, verbose=True)

# Verify results
print(f"\nTotal predictions received: {len(received_predictions)}")
print(f"Predictions: {received_predictions}")
```
